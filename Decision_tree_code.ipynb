```python
# Import all libraries which is related to your output
import pandas as pd # pandas we use for read the output
from sklearn.tree import DecisionTreeClassifier # scrikit-learn library used in machine learning
# and it draw decision tree and how to tree make decison as per your output
```

```python

# Create own dataset and variable name = data
data = {
    "Weather": ["Sunny", "Cloudy", "Sunny", "Cloudy", "Rainy", "Rainy", "Rainy", "Sunny", "Cloudy", "Rainy"],
    "Temperature": ["Hot", "Hot", "Mild", "Mild", "Mild", "Cool", "Mild", "Hot", "Hot", "Mild"],
    "Humidity": ["High", "High", "Normal", "High", "High", "Normal", "High", "High", "Normal", "High"],
    "Wind": ["Weak", "Weak", "Strong", "Strong", "Strong", "Strong", "Weak", "Strong", "Weak", "Strong"],
    "Play": ["No", "Yes", "Yes", "Yes", "No", "No", "Yes", "No", "Yes", "No"]
}
```

```python
# here df means dataframe to represent ur data in form of row and columns
df = pd.DataFrame(data)
df
```

```python

# Convert text data to Numeric data beacuse machine can't understand ur text data
# this all process name is Label Encoding
# repalce means text to numeric
df_encoded = df.replace({"Sunny":0, "Cloudy":1, "Rainy":2,
                        "Hot":0, "Mild":1, "Cool":2,
                        "High":0, "Normal":1,
                        "Weak":0, "Strong":1,
                        "No":0, "Yes":1})
```

```python
df_encoded # for print your data
```

```python
# Separate input and output
X = df_encoded.drop("Play", axis=1) # here u can drop play attribute and axis = 1 means remove column (not rrow)
y = df_encoded["Play"] # here u can use play aattribute as a y
```

```python
X
```

```python
y
```

```python
# Train Decicion Tree model
model = DecisionTreeClassifier() # here u can fir model , desicion tree model is totally blank
model.fit(X,y) # here u can fir your data in the form of x and y
# x = Weather, Temperature, Humidity, Wind
#y = output labels
```

```python
print("\n Model Trained") # it's shows your model is trained for prediction
# because u fit here in the basis of x and y
```

```python
# Make a sample prediction
sample = [[1,1,0,0]] # here u can create ur own sample data for predicition
prediction = model.predict(sample)
```

```python
if prediction[0] == 1:
  result = "play"
else:
  result = "Don't Play"

print(result)
```
